## Section 5.1 — Data: MNIST

Load the first 1 000 MNIST training images, standardise, and reduce to 30 principal components — the standard preprocessing for t-SNE on MNIST.

```python
X, y = load_mnist(n_samples=1000, pca_components=30)
print(f"X shape: {X.shape}  |  classes: {sorted(set(y.tolist()))}")
```

## Section 5.2 — t-SNE Embeddings under Each Initialisation Strategy

Run `tsne_eigen` with each of the four initialisation methods.
Set `T=3` to visualise only the starting configuration (no gradient descent).
Set `T=1000` for the fully-optimised embedding.

```python
# Set initialization_method to one of: "random", "PCA", "spectral", "laplacian"
# Set T=3 to see only the initialisation (no gradient descent steps).
method = "spectral"
T = 3  # increase to 500 or 1000 for a fully-optimised embedding

Y = tsne_eigen(X, perplexity=30, T=T, eta=500,
               initialization_method=method)

plot_embedding(Y, y, title=f"MNIST — {method} initialisation (T={T})")
```
